### Problem

Input:
start_times = [1,3,5]
end_times = [2,4,6]

Output is 1.

### Logic

1. Sort input lists, so they are both from low to high.
2. For each start_time, compare it to end_times[current_meeting_index].
3. If start < end_times[current_meeting_index], used_rooms+1.
4. If start >= end_times[current_meeting_index], no new room needed; current_meeting_index+1.

This way we won't need to build meeting room pairs ([1,2], [3,4], [5,6]) and also no need to use max logic since it will be added toward used_rooms when needed.

### Code

```python
def min_meeting_rooms(start_times, end_times):
    # 1: sort
    start_times.sort()
    end_times.sort()

    used_rooms = 0
    current_meeting_index = 0

    # 2-4: main logic, loop through meetings
    for start in start_times:
        if start < end_times[current_meeting_index]:
            used_rooms += 1
        else:
            current_meeting_index += 1

    return used_rooms
```
